# Human-review alert workflow
Synthetic demonstration — not operational validation.


```python
from iot_pm.simulation.generator import simulate
from iot_pm.schemas import SensorReading
from iot_pm.ml.alerts import score_rules
import pandas as pd
df=simulate(120)
alerts=[]
for row in df.to_dict('records'):
    event=SensorReading(**row)
    a=score_rules(event)
    if a.severity!='low': alerts.append(a.model_dump())
print('Flagged events:',len(alerts))
print(pd.DataFrame(alerts).head())
print('Alerts are not automatic work orders; human review required.')
```
