# Machine-specific sensor profiles
Synthetic demonstration — not operational validation.


```python
from iot_pm.simulation.generator import simulate
import matplotlib.pyplot as plt
df=simulate(1100,machines=5)
for machine,g in df.groupby('machine_id'):
    plt.plot(g.operating_hours,g.vibration_mm_s,label=machine)
plt.xlabel('Operating hours');plt.ylabel('Vibration mm/s');plt.legend();plt.show()
```
