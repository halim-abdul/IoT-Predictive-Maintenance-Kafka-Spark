# Alert threshold and recall tradeoffs
Synthetic demonstration — not operational validation.


```python
from iot_pm.ml.evaluation import select_threshold,classification_at_threshold
from sklearn.metrics import precision_recall_curve
import numpy as np
import matplotlib.pyplot as plt
rng=np.random.default_rng(11)
y=rng.binomial(1,0.12,1000);p=np.clip(0.2+0.5*y+rng.normal(0,0.24,1000),0,1)
t=select_threshold(y,p,min_recall=.8)
print('Threshold:',t,'Metrics:',classification_at_threshold(y,p,t))
prec,rec,_=precision_recall_curve(y,p)
plt.plot(rec,prec);plt.xlabel('Recall');plt.ylabel('Precision');plt.show()
```
